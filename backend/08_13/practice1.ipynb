````python
```python
import os
import psycopg
from dotenv import load_dotenv

load_dotenv()
conninfo = os.getenv("DATABASE_URL")

with psycopg.connect(conninfo) as conn:      # with 結束自動關連線 + commit
    with conn.cursor() as cur:               # with 結束自動關游標
        cur.execute("SELECT version();")
        print(cur.fetchall())
```
````
